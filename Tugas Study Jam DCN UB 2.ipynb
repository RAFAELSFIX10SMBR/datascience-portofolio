## 1.2 For Loop - Si Pekerja Terencana

**For Loop** dipakai kalau kita **sudah tahu** berapa kali harus jalan, atau punya daftar yang jelas.

### Sintaks Dasar:
```python
for variable_sementara in koleksi_data:
    # lakukan sesuatu
```

````python
# === CONTOH 1: Iterasi List Sederhana ===
keranjang_buah = ["Apel", "Jeruk", "Mangga", "Pisang", "Anggur"]

print(" Isi keranjang belanja:")
for buah in keranjang_buah:
    print(f"  - {buah}")

# CARA BACA:
# 'for buah in keranjang_buah' = Untuk setiap item (dinamai 'buah') di dalam keranjang_buah
# Setiap putaran, variabel 'buah' berisi nilai berbeda (Apel, Jeruk, dst)
````

````python
# === CONTOH 2: Loop dengan Range ===
# range(start, stop, step) - menghasilkan urutan angka

print(" Contoh range(5):")
for i in range(5):  # 0, 1, 2, 3, 4 (berhenti SEBELUM 5)
    print(f"  Iterasi ke-{i}")

print("\n Contoh range(1, 6):")
for i in range(1, 6):  # 1, 2, 3, 4, 5 (mulai dari 1, berhenti sebelum 6)
    print(f"  Angka: {i}")

print("\n Contoh range(10, 0, -2):")
for i in range(10, 0, -2):  # 10, 8, 6, 4, 2 (mundur 2 langkah)
    print(f"  Hitung mundur: {i}")
````

````python
# === CONTOH 3: CASE DATA SCIENCE - Menghitung Statistik Dasar ===
# Skenario: Kamu punya data penjualan harian toko online

penjualan_harian = [150000, 230000, 180000, 420000, 310000, 275000, 190000]
hari = ["Senin", "Selasa", "Rabu", "Kamis", "Jumat", "Sabtu", "Minggu"]

# Menghitung total penjualan dengan loop
total = 0
for nominal in penjualan_harian:
    total = total + nominal  # atau: total += nominal

print(f" Total penjualan seminggu: Rp {total:,}")
print(f" Rata-rata per hari: Rp {total // len(penjualan_harian):,}")

# PENJELASAN:
# Kita siapkan 'total = 0' sebagai wadah kosong
# Setiap loop, kita tambahkan satu nominal ke wadah
# Format {:,} menambahkan pemisah ribuan (150,000)
````

````python
# === CONTOH 4: Enumerate - Dapat Index dan Value Sekaligus ===
# Sangat berguna ketika butuh nomor urut!

produk = ["Laptop", "Mouse", "Keyboard", "Monitor", "Webcam"]

print(" Daftar Inventaris:")
for nomor, nama in enumerate(produk, start=1):  # start=1 biar mulai dari 1, bukan 0
    print(f"  {nomor}. {nama}")

# CARA BACA enumerate():
# enumerate menghasilkan pasangan (index, value)
# Kita 'unpack' jadi dua variabel: nomor dan nama
````

````python
# === CONTOH 5: CASE DATA SCIENCE - Mencari Outlier ===
# Skenario: Deteksi data yang mencurigakan (terlalu tinggi/rendah)

transaksi = [25000, 30000, 28000, 950000, 27000, 31000, 5000, 29000]
batas_atas = 100000
batas_bawah = 10000

print(" Deteksi Transaksi Anomali:")
for idx, nilai in enumerate(transaksi):
    if nilai > batas_atas:
        print(f"   Index {idx}: Rp {nilai:,} - TERLALU TINGGI!")
    elif nilai < batas_bawah:
        print(f"   Index {idx}: Rp {nilai:,} - TERLALU RENDAH!")

# Di Data Science, ini namanya Outlier Detection
# Outlier bisa jadi error input atau memang transaksi abnormal
````

## 1.3 While Loop - Si Pekerja Bersyarat

**While Loop** jalan terus **selama kondisi masih True**. Baru berhenti kalau kondisinya jadi False.

###  Analogi Game
While loop itu seperti boss fight di game: "Selama HP boss masih > 0, terus serang!"

### Sintaks Dasar:
```python
while kondisi:
    # lakukan sesuatu
    # PENTING: harus ada yang mengubah kondisi, atau infinite loop!
```

###  PERINGATAN PENTING
While loop bisa menyebabkan **infinite loop** (loop tanpa akhir) kalau kondisinya tidak pernah berubah jadi False. Kalau terjadi, tekan **Ctrl+C** atau restart kernel.

````python
# === CONTOH 6: While Loop Dasar ===
counter = 5

print(" Hitung mundur peluncuran roket:")
while counter > 0:
    print(f"  {counter}...")
    counter = counter - 1  # WAJIB! Ini yang mengubah kondisi

print("   LIFTOFF!")

# Tanpa 'counter = counter - 1', loop akan jalan selamanya!
````

````python
# === CONTOH 7: CASE DATA SCIENCE - Simulasi Pertumbuhan ===
# Skenario: Berapa lama investasi mencapai target?

modal_awal = 1000000  # Rp 1 juta
target = 2000000      # Rp 2 juta
bunga_tahunan = 0.08  # 8% per tahun

modal = modal_awal
tahun = 0

while modal < target:
    modal = modal * (1 + bunga_tahunan)  # Compound interest
    tahun += 1
    print(f"  Tahun {tahun}: Rp {modal:,.0f}")

print(f"\n Butuh {tahun} tahun untuk mencapai Rp {target:,}")
````

````python
# === CONTOH 8: CASE DATA SCIENCE - Iterasi Sampai Konvergen ===
# Skenario: Banyak algoritma ML jalan sampai error cukup kecil

error = 100
threshold = 0.01
iterasi = 0

print(" Simulasi Training Model:")
while error > threshold:
    error = error * 0.7  # Error berkurang 30% tiap iterasi (simulasi)
    iterasi += 1
    print(f"  Iterasi {iterasi}: Error = {error:.4f}")

print(f"\n Model konvergen setelah {iterasi} iterasi!")

# Ini mirip konsep 'convergence' di Machine Learning
# Model berhenti training kalau improvement sudah minimal
````

## 1.4 Break, Continue, dan Pass

Tiga keyword untuk mengontrol alur loop:

| Keyword | Fungsi |
|---------|--------|
| `break` | Keluar dari loop sepenuhnya |
| `continue` | Skip iterasi saat ini, lanjut ke berikutnya |
| `pass` | Tidak melakukan apa-apa (placeholder) |

````python
# === CONTOH 9: Break - Berhenti di Tengah Jalan ===
# Skenario: Cari customer pertama yang punya hutang

customers = [
    {"nama": "Andi", "hutang": 0},
    {"nama": "Budi", "hutang": 0},
    {"nama": "Caca", "hutang": 150000},
    {"nama": "Dedi", "hutang": 200000},  # Ini tidak akan dicek
]

print(" Mencari customer dengan hutang...")
for customer in customers:
    if customer["hutang"] > 0:
        print(f"   Ditemukan: {customer['nama']} - Hutang Rp {customer['hutang']:,}")
        break  # Langsung keluar, tidak perlu cek sisanya
    print(f" {customer['nama']} tidak punya hutang")
````

````python
# === CONTOH 10: Continue - Skip Data Tertentu ===
# Skenario: Proses hanya data yang valid

data_sensor = [23.5, None, 25.1, -999, 24.8, None, 26.2]
# None = sensor error, -999 = missing value

print(" Data Suhu Valid:")
for nilai in data_sensor:
    if nilai is None or nilai < 0:
        continue  # Skip data invalid, lanjut ke berikutnya
    print(f"  Suhu: {nilai}°C")

# Data cleaning basic: skip/ignore data yang tidak valid
````

## 1.5 Nested Loop (Loop Bersarang)

Loop di dalam loop. Berguna untuk data 2 dimensi (tabel/matrix).

###  Analogi
Kamu punya 3 loyang pizza, masing-masing dipotong 8 slice. Untuk makan semuanya:
- Loop luar: Untuk setiap loyang...
- Loop dalam: Untuk setiap slice di loyang itu...

````python
# === CONTOH 11: Nested Loop - Tabel Perkalian ===
print(" Tabel Perkalian 1-5:")
print()

for i in range(1, 6):  # Baris
    for j in range(1, 6):  # Kolom
        hasil = i * j
        print(f"{hasil:4}", end="")  # end="" supaya tidak ganti baris
    print()  # Ganti baris setelah satu row selesai
````

````python
# === CONTOH 12: CASE DATA SCIENCE - Proses Data Tabel ===
# Skenario: Data penjualan per cabang per bulan

penjualan = [
    [100, 120, 150],  # Cabang Jakarta: Jan, Feb, Mar
    [90, 110, 130],   # Cabang Bandung
    [80, 95, 115],    # Cabang Surabaya
]
cabang = ["Jakarta", "Bandung", "Surabaya"]
bulan = ["Jan", "Feb", "Mar"]

print(" Laporan Penjualan (dalam juta):")
print()

for i, kota in enumerate(cabang):
    total_cabang = 0
    print(f"  {kota}:")
    for j, bln in enumerate(bulan):
        nilai = penjualan[i][j]
        total_cabang += nilai
        print(f"    {bln}: {nilai}M")
    print(f"    Total: {total_cabang}M")
    print()
````

## 1.6 List Comprehension - Loop dalam 1 Baris!

Python punya cara singkat untuk membuat list baru dari list yang sudah ada. Ini **SANGAT PENTING** di Data Science karena lebih cepat dan lebih mudah dibaca.

### Sintaks:
```python
list_baru = [ekspresi for item in iterable if kondisi]
```

````python
# === CONTOH 13: List Comprehension Dasar ===

# Cara lama dengan for loop:
angka = [1, 2, 3, 4, 5]
kuadrat_lama = []
for x in angka:
    kuadrat_lama.append(x ** 2)
print(f"Cara lama: {kuadrat_lama}")

# Cara baru dengan list comprehension:
kuadrat_baru = [x ** 2 for x in angka]
print(f"List comprehension: {kuadrat_baru}")

# Sama hasilnya, tapi jauh lebih singkat!
````

````python
# === CONTOH 14: List Comprehension dengan Kondisi ===

nilai_ujian = [85, 42, 78, 55, 92, 38, 67, 71]

# Ambil hanya yang lulus (>= 60)
yang_lulus = [n for n in nilai_ujian if n >= 60]
print(f"Yang lulus: {yang_lulus}")

# Ambil yang tidak lulus
yang_remedial = [n for n in nilai_ujian if n < 60]
print(f"Yang remedial: {yang_remedial}")
````

````python
# === CONTOH 15: CASE DATA SCIENCE - Feature Engineering ===
# Skenario: Membuat fitur baru dari data yang ada

harga = [100000, 250000, 75000, 500000, 180000]

# Buat kategori harga
kategori = ["Mahal" if h > 200000 else "Murah" for h in harga]
print(f"Harga: {harga}")
print(f"Kategori: {kategori}")

# Normalisasi ke range 0-1
min_h = min(harga)
max_h = max(harga)
normalized = [(h - min_h) / (max_h - min_h) for h in harga]
print(f"Normalized: {[round(n, 2) for n in normalized]}")
````

````python
# === CONTOH 16: CASE DATA SCIENCE - Data Cleaning ===
# Skenario: Bersihkan dan transformasi data

nama_kotor = ["  andi  ", "BUDI", "  CaCa", "deDI  ", None, ""]

# Bersihkan: strip, lowercase, filter yang kosong/None
nama_bersih = [
    nama.strip().lower()
    for nama in nama_kotor
    if nama is not None and nama.strip() != ""
]

print(f"Data kotor: {nama_kotor}")
print(f"Data bersih: {nama_bersih}")
````

Map Function - Mesin Transformasi Otomatis

**Map** adalah cara instan untuk menjalankan sebuah fungsi ke SELURUH anggota list tanpa nulis loop manual.

###  Analogi Mesin Fotokopi Canggih
Bayangkan mesin fotokopi yang bisa sekaligus:
1. Mengecilkan ukuran kertas
2. Mengubah warna jadi hitam putih
3. Menambahkan watermark

Kamu masukkan setumpuk 100 kertas, keluar 100 kertas yang sudah diproses. **Itulah map!**

### Sintaks:
```python
hasil = map(fungsi, koleksi_data)
# Jangan lupa convert ke list: list(hasil)
```

````python
# === CONTOH 1: Map dengan Fungsi Biasa ===

def tambah_ppn(harga):
    """Tambahkan PPN 11% ke harga"""
    return harga * 1.11

daftar_harga = [10000, 25000, 50000, 75000, 100000]

# Terapkan fungsi ke semua item
harga_dengan_pajak = list(map(tambah_ppn, daftar_harga))

print(" Harga Asli vs Setelah PPN:")
for asli, pajak in zip(daftar_harga, harga_dengan_pajak):
    print(f"  Rp {asli:,} → Rp {pajak:,.0f}")

# KENAPA PAKAI list()?
# map() mengembalikan 'map object' (iterator), bukan list langsung
# Kita wrap dengan list() supaya hasilnya jadi list yang bisa diakses
````

````python
# === CONTOH 2: Lambda - Fungsi Instan Tanpa Nama ===

# Daripada bikin fungsi terpisah buat operasi sederhana...
# Pake lambda aja!

angka = [1, 2, 3, 4, 5]

# Lambda buat kuadrat
kuadrat = list(map(lambda x: x ** 2, angka))
print(f"Kuadrat: {kuadrat}")

# Lambda buat kelipatan 3
kali_tiga = list(map(lambda x: x * 3, angka))
print(f"Kali 3: {kali_tiga}")

# CARA BACA LAMBDA:
# lambda x: x ** 2
# 'lambda' = keyword buat bikin fungsi instan
# 'x' = parameter input
# 'x ** 2' = hasil yang dikembalikan
````

````python
# === CONTOH 3: CASE DATA SCIENCE - Normalisasi Data ===
# Skenario: Konversi suhu dari Celsius ke Fahrenheit

suhu_celsius = [0, 20, 25, 30, 37, 100]

# Formula: F = C * 9/5 + 32
suhu_fahrenheit = list(map(lambda c: c * 9/5 + 32, suhu_celsius))

print("Konversi Suhu:")
for c, f in zip(suhu_celsius, suhu_fahrenheit):
    print(f"  {c} C = {f:.1f} F")
````


## 2.2 Filter Function - Penyaring Data Otomatis

**Filter** milih item yang memenuhi kondisi tertentu.

### Analogi
Kayak saringan pasir. Kamu tuang campuran pasir dan kerikil, yang lolos cuma butiran pasir kecil.

### Sintaks:
```python
hasil = filter(fungsi_boolean, koleksi_data)
# Fungsi harus return True atau False
```


````python
# === CONTOH 4: Filter Dasar ===

nilai_ujian = [85, 42, 78, 55, 92, 38, 67, 71, 88, 45]

# Filter yang lulus (>= 60)
lulus = list(filter(lambda x: x >= 60, nilai_ujian))

# Filter yang nggak lulus
tidak_lulus = list(filter(lambda x: x < 60, nilai_ujian))

# len adalah method yang digunakan untuk mengecek panjang dari isi suatu variable. misal len(nilai_ujian), karna jumlah isi dari variable tersebut ada 10 item, maka akan memunculkan angka 10 di total siswa
print(f"Total siswa: {len(nilai_ujian)}")
print(f"Lulus ({len(lulus)} orang): {lulus}")
print(f"Nggak lulus ({len(tidak_lulus)} orang): {tidak_lulus}")
````
