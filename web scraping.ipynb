```python


driver = webdriver.Chrome()  # or Firefox()
driver.get("https://www.kickavenue.com/search?keyword=court+vision&sort_by=most_popular")

# Wait for products to load
WebDriverWait(driver, 10).until(
    EC.presence_of_element_located((By.CLASS_NAME, "img-sneaker"))
)

soup = BeautifulSoup(driver.page_source, "html.parser")
shoes = soup.find_all("div", class_="img-sneaker card-body")

for shoe in shoes:
    brand = shoe.find("span", class_="price-tag brand")
    title = shoe.find("h3", class_="price-tag")
    price = shoe.find("span", class_="currency-tag")

    if brand and title and price:
        print(brand.text.strip(), title.text.strip(), price.text.strip())

driver.quit()
```
