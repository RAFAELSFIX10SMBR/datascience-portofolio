# BAGIAN 1: Pemanasan (The Basics)
Sebelum lari maraton, kita stretching dulu. Jangan remehkan dasar, karena kalau pondasi rapuh, kodinganmu bakal rubuh.

## 1. Variable & Tipe Data
Variable itu cuma label/nama yang kita tempel ke sebuah nilai. Python itu *dynamically typed*, jadi dia pinter nebak tipe datanya sendiri.

**Tipe Data Wajib Tau:**
* `str` (String): Teks. Pembungkusnya kutip satu `'` atau dua `"`.
* `int` (Integer): Bilangan bulat. Contoh: Jumlah mantan.
* `float` (Float): Bilangan desimal. Contoh: IPK, Saldo ATM.
* `bool` (Boolean): Kebenaran mutlak. Cuma `True` atau `False`.

```python
# Deklarasi Variable
nama_karyawan = "Budi"
gaji_pokok = 8000000
bonus_lembur = 500000.50
is_permanent = True

# Cek Tipe Data (Penting buat debugging)
print(f"Tipe data gaji: {type(gaji_pokok)}")

# Casting (Maksa ubah tipe data)
# Use Case: Input dari user itu selalu String, harus diubah ke Int buat dihitung
umur_input = "25"
umur_angka = int(umur_input)
print(f"Umur tahun depan: {umur_angka + 1}")
```

**Tips & Trick:**
* Gunakan `snake_case` untuk nama variable (huruf kecil semua dipisah underscore). Jangan `camelCase` kayak programmer sebelah.
* Nama variable harus deskriptif. `x` itu membingungkan, `harga_barang` itu jelas.

**Common Mistakes:**
* `NameError`: Manggil variable yang belum dibuat (biasanya typo).
* `TypeError`: Menjumlahkan String sama Integer. Minyak sama air gak bisa nyatu, bos.

## 2. Flow Control (Logika Percabangan)
Biar program gak lempeng-lempeng aja, kita kasih otak. Kuncinya di `if`, `elif`, dan `else`.

**Operator Logika:**
* `and`: Dua-duanya harus benar.
* `or`: Salah satu benar, gas.
* `not`: Kebalikan.
* `in`: Cek apakah ada di dalam list/string.

```python
nilai_coding = 80
nilai_softskill = 70
blacklist_hr = False

# Logika Seleksi Karyawan
if blacklist_hr:
    print("Ditolak: Blacklist HR.")
elif nilai_coding > 75 and nilai_softskill > 75:
    print("Diterima: Kandidat Emas.")
elif nilai_coding > 75 or nilai_softskill > 80:
    print("Diterima: Kandidat Perak (Perlu Training).")
else:
    print("Ditolak: Skill belum cukup.")
```

**Common Mistakes:**
* Lupa titik dua `:` di akhir baris `if`.
* **IndentationError**: Spasi di baris bawahnya gak rata. Python itu perfeksionis soal kerapian.

## 3. Functions (Mesin Pabrik)
Kalau kamu nulis kode yang sama lebih dari dua kali, bungkus jadi fungsi. Prinsip DRY (Don't Repeat Yourself).

Penting:
* `return`: Mengembalikan nilai hasil proses.
* `default parameter`: Nilai cadangan kalau user lupa ngisi input.

```python
# Default parameter: tax defaultnya 10% kalau gak diisi
def hitung_gaji_bersih(gaji_kotor, tax=0.1):
    potongan = gaji_kotor * tax
    bersih = gaji_kotor - potongan
    return bersih

gaji_andi = hitung_gaji_bersih(10000000)
gaji_budi = hitung_gaji_bersih(20000000, tax=0.2) # Pajak orang kaya beda

print(f"Gaji Andi: {gaji_andi}, Gaji Budi: {gaji_budi}")
```

# BAGIAN 2: Data Structures (Gudang Data)
Ini senjata utama Data Scientist. Kita harus tau kapan pakai List, Tuple, Set, atau Dictionary.

## 1. List
List itu urut (ordered) dan bisa diubah (mutable). Isinya bisa campur aduk.

**Method Paling Sering Dipakai:**
* `.append(x)`: Nambah `x` di paling belakang.
* `.pop(index)`: Hapus dan ambil data di posisi `index`.
* `.sort()`: Mengurutkan data (permanen).
* `.extend(list_lain)`: Menggabungkan dua list.
* `len(list)`: Menghitung jumlah data.

```python
transaksi = [150, 200, 50, 300]

# 1. Nambah Data
transaksi.append(500)

# 2. Slicing (Memotong Data) -> [Start : Stop : Step]
# Ambil 3 data terakhir
print(f"3 Transaksi Terakhir: {transaksi[-3:]}")

# 3. Modifikasi
transaksi[0] = 999 # Revisi data pertama

# 4. Sorting
transaksi.sort(reverse=True) # Urut dari besar ke kecil
print(f"Sorted: {transaksi}")
```

**Common Mistakes:**
* `IndexError`: Minta data index ke-10 padahal datanya cuma 5.
* Modifikasi list saat sedang di-looping (Ini bahaya, hasilnya bisa acak-acakan).

## 2. Dictionary
Penyimpanan berbasis `Key: Value`. Akses datanya super cepat, gak perlu nyari urut satu-satu.

**Method Penting:**
* `.keys()`: Ambil semua kuncinya.
* `.values()`: Ambil semua nilainya.
* `.get(key, default)`: Ambil nilai dengan aman (gak error kalau key gak ada).
* `.update(dict_lain)`: Gabungin dictionary.

```python
profil_user = {
    "id": 101,
    "username": "data_ninja",
    "skills": ["Python", "SQL"]
}

# Akses Aman
# print(profil_user["alamat"]) # INI BAKAL ERROR (KeyError)
alamat = profil_user.get("alamat", "Alamat tidak diketahui") # INI AMAN

# Nambah Key Baru
profil_user["level"] = "Senior"

print(f"User Level: {profil_user['level']}")
print(f"Alamat: {alamat}")
```

## 3. Set (Himpunan Unik)
Anti duplikat. Kalau ada data kembar, langsung dibuang satu.

**Use Case:**
* Menghitung jumlah pengunjung unik (Unique Visitors).
* Mencari irisan (Intersection) antar dataset.

```python
id_pengunjung = [1, 2, 3, 1, 2, 4, 5]

# Buang duplikat
unik = set(id_pengunjung)
print(f"Pengunjung Unik: {unik}")

# Operasi Himpunan
karyawan_a = {"Andi", "Budi", "Caca"}
karyawan_b = {"Budi", "Dedi", "Eka"}

# Siapa yang ada di kedua tim? (Intersection)
double_job = karyawan_a.intersection(karyawan_b)
print(f"Double Job: {double_job}")
```

# BAGIAN 3: Comprehensions & Lambda
Fitur ini yang bikin programmer Python kelihatan keren. Nulis kode 5 baris jadi 1 baris (One-Liner).

## 1. List Comprehension
Membuat list baru dari list lama dengan sintaks ringkas.

**Rumus:** `[ekspresi for item in list if kondisi]`

```python
harga_usd = [10, 20, 30, 40]
kurs = 15000

# Cara Lama (Cupu)
# harga_idr = []
# for h in harga_usd:
#     harga_idr.append(h * kurs)

# Cara Ninja (Suhu)
harga_idr = [h * kurs for h in harga_usd]
print(f"Harga IDR: {harga_idr}")

# Pakai Filter: Cuma ambil yang harganya di atas 20 USD
mahal = [h for h in harga_usd if h > 20]
print(f"Barang Mahal: {mahal}")
```

## 2. Lambda Function
Fungsi anonim (tanpa nama) yang biasanya cuma dipakai sekali. Sering banget dipakai bareng `map()`, `filter()`, atau `sort()`.

**Use Case:** Custom sorting.

```python
data_siswa = [
    {"nama": "Zaki", "nilai": 80},
    {"nama": "Ali", "nilai": 90},
    {"nama": "Budi", "nilai": 85}
]

# Sorting List of Dictionary berdasarkan KEY tertentu ('nilai')
data_siswa.sort(key=lambda x: x['nilai'], reverse=True)

print("Ranking Siswa:", data_siswa)
```

### **LATIHAN MINI 2: FILTER DATA**
Ada list email:
`emails = ["budi@gmail.com", "andi@yahoo.com", "siti@gmail.com", "dedi@hotmail.com"]`

**Misi:**
Ambil hanya email yang menggunakan domain `@gmail.com` menggunakan **List Comprehension**.

*Clue: String method `endswith()` atau cek string `in` string.*

```python
# Code Latihan 2 di sini

emails = ["budi@gmail.com", "andi@yahoo.com", "siti@gmail.com", "dedi@hotmail.com"]

gmail = [e for e in emails if "@gmail.com" in e]
gmail2 = [e for e in emails if e.endswith("@gmail.com")]
print(gmail)
print(gmail2)
```

# BAGIAN 4: Operasi Plastik Data (String Manipulation)
Data asli itu jarang bersih. Pasti ada spasi berlebih, huruf besar kecil ngawur, atau karakter aneh. Ini skill wajib buat *Data Cleaning*.

**Method Wajib Hafal:**
* `.strip()`: Hapus spasi di depan & belakang.
* `.lower()` / `.upper()`: Ubah kapitalisasi.
* `.replace(old, new)`: Ganti kata/karakter.
* `.split(delimiter)`: Pecah kalimat jadi list kata.
* `.join(list)`: Gabungin list jadi kalimat.

```python
teks_alay = "  PrOmo DisKON 50% haNYA haRI iNi!!!  "

# 1. Bersihin spasi
bersih = teks_alay.strip()

# 2. Jadiin huruf kecil semua (Normalisasi)
bersih = bersih.lower()

# 3. Hapus tanda seru
bersih = bersih.replace("!", "")

# 4. Pecah jadi kata-kata (Tokenization)
kata_kata = bersih.split(" ")

print(f"Teks Asli: '{teks_alay}'")
print(f"Teks Bersih: '{bersih}'")
print(f"Token: {kata_kata}")
```

**Tips: F-String Formatting**
Biar output angka enak dilihat, pake formatting.

```python
profit = 12500000.5678
persen = 0.856

# ,.2f artinya: kasih pemisah ribuan (,) dan ambil 2 angka belakang koma (.2f)
print(f"Profit: Rp{profit:,.2f}")
print(f"Akurasi: {persen:.1%}")
```

### **LATIHAN MINI 3: TEXT NORMALIZER**
Diberikan list kode produk yang berantakan:
`codes = [" PROD-001 ", "prod-002", "PROD-003 "]`

**Misi:**
Bersihkan data sehingga menjadi: `['PROD-001', 'PROD-002', 'PROD-003']`
Semua harus huruf besar dan tanpa spasi.

*Clue: Kombinasikan List Comprehension dengan `.strip()` dan `.upper()`.*

```python
# Code Latihan 3 di sini
codes = [" PROD-001 ", "prod-002", "PROD-003 "]

codes_bersih = [c.strip().upper() for c in codes]
print(codes_bersih)
```

# BAGIAN 5: Detektif Error (Debugging)
Jangan panik kalau merah. Error message itu petunjuk, bukan vonis mati.

**Jenis Error Paling Sering Muncul:**
1.  **SyntaxError**: Typo, kurang kurung, kurang titik dua.
2.  **IndentationError**: Spasi gak rata.
3.  **NameError**: Manggil variable yang belum dibuat.
4.  **TypeError**: Operasi antar tipe data yang gak cocok.
5.  **IndexError**: Minta data di luar batas list.
6.  **KeyError**: Minta kunci yang gak ada di Dictionary.

**Tips Debugging:**
* Baca baris paling bawah dari pesan error.
* Gunakan `print()` di tengah kode untuk cek isi variable.
* Gunakan `try-except` untuk menangani error yang bisa diprediksi.

```python
data_palsu = {"harga": 5000}

try:
    print(data_palsu["stok"])
except KeyError:
    print("Waduh, kuncinya gak ada bos! Pake .get() aja biar aman.")
except Exception as e:
    print(f"Error lain: {e}")
```
